```python
#9
product_analysis = data.groupby('Product Name').agg(
    purchase_count=('Product Name', 'count'),
    revenue=('Total Sales', 'sum')
)
product_analysis
```

```python
m1 = product_analysis['purchase_count'].median()
m2 = product_analysis['revenue'].median()

freq_low = product_analysis[(product_analysis['purchase_count']>m1) & (product_analysis['revenue']<m2)]
freq_low.sort_values(['purchase_count','revenue'],ascending=False)
```

```python
freq_high = product_analysis[(product_analysis['purchase_count']<m1)&(product_analysis['revenue']>m2)]
freq_high.sort_values(['purchase_count','revenue'],ascending=False)
```
